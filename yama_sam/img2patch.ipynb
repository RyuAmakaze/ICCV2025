## Img&Npy -> 64*64[pix] patch image

```python
import os
import numpy as np
import cv2

# 入力パス
IMAGE_PATH = "images/test_hihu.jpg"
NPY_PATH = os.path.join("output", os.path.basename(IMAGE_PATH) + ".npy")

# 出力ディレクトリ
output_dir = os.path.splitext(os.path.basename(IMAGE_PATH))[0]
os.makedirs(output_dir, exist_ok=True)

# パラメータ
PATCH_SIZE = 64
OVERLAP = PATCH_SIZE // 2  # 50%のオーバーラップ
```

```python
# データ読み込み
npy = np.load(NPY_PATH)
img = cv2.imread(IMAGE_PATH)

# 画像サイズ取得
h, w, _ = img.shape

# パッチのスライド処理
for y in range(0, h - PATCH_SIZE + 1, OVERLAP):
    for x in range(0, w - PATCH_SIZE + 1, OVERLAP):
        patch_img = img[y:y + PATCH_SIZE, x:x + PATCH_SIZE]
        patch_npy = npy[y:y + PATCH_SIZE, x:x + PATCH_SIZE]

        # True が含まれているパッチのみ保存
        if np.any(patch_npy):
            patch_filename = f"{x}_{y}_{x + PATCH_SIZE}_{y + PATCH_SIZE}.png"
            cv2.imwrite(os.path.join(output_dir, patch_filename), patch_img)

print(f"パッチ画像を {output_dir} に保存しました。")
```
